```python
# 2. Perform a Left Merge between Titanic and Insurance.
left = pd.merge(
    df,
    insurance,
    on ='PassengerId',
    how ='left'
)
left
```

```python
# 3. Perform a Right Merge between Titanic and Insurance.
right = pd.merge(
    df,
    insurance,
    on ='PassengerId',
    how ='right'
)
right
```

```python
# 1. Perform vertical concatenation of January and February bookings.
vartical = pd.concat([jan ,feb])
vartical
```

```python
#2. Perform vertical concatenation using ignore_index=True .
vartical = pd.concat([jan,feb],ignore_index = True)
vartical
```

```python
#3. Perform horizontal concatenation of bookings and city details.
horizontal = pd.concat([vartical,city],axis = 1)
horizontal
```

```python
# 3. Create an Age_Group column using apply() .
import pandas as pd
df1 = pd.read_csv('/kaggle/input/datasets/yasserh/titanic-dataset/Titanic-Dataset.csv')
```

```python
def age_group(Age):
    if pd.isnull(Age):
        return 'unknown'
    elif Age < 18:
        return "Child"
    elif Age < 60:
        return 'Adult'
    else:
        return 'Senior'

df1["Age_Group"] = df1['Age'].apply(age_group)
df1
```

```python
# 4. Create a Gender_Code column using map() .
df1['Gender_Code'] = df1['Sex'].map({
    'male' : 0,
    'female' : 1 
})
```

```python
df1
```

```python
#5. Increase every value in a sample numeric DataFrame using applymap() .
df2 = pd.DataFrame({
    'StId': [101,108,140,120],
    'Age' : [12,17,19,20]
})
df2 = df2.applymap(lambda x : x*2)
df2
```

```python
# 6. Create a Premium_Passenger column using a lambda function.
#df3 = pd.read_csv('/kaggle/input/datasets/yasserh/titanic-dataset/Titanic-Dataset.csv')
df1['Premium_Passenger'] = df1['Pclass'].apply(lambda x: 'Yes' if x == 1 else 'No')
df1
```

```python
# 7. Create a Family_Size feature.
df1['Family_Size'] = df1['SibSp'] + df1['Parch'] + 1
```

```python
df1
```

```python
# 8. Create an Is_Alone feature.
df1[' Is_Alone'] = (df1['Family_Size'] == 1).astype(int)
df1
```

```python
#9. Create a Fare_Per_Person feature.
df1['Fare_Per_Person'] =df1["Fare"] / df1['Family_Size']
df1
```

```python
#10. Create a Fare_Category feature using bins.
```

```python
df1['Fare_Category'] = pd.cut(
    df1['Fare'],
    bins=[0, 10, 25, 600],
    labels=['Low', 'Medium', 'High'],
    include_lowest=True
)

df1
```
